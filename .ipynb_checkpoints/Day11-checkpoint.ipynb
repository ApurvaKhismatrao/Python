# Q1. Program to find the reverse of a user-entered number

```python

num = int(input("Enter a number: "))
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

print("Reverse number is:", rev)
```

# Q2. Program to check whether a number is a palindrome or not

```python

num = int(input("Enter a number: "))
temp = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

if temp == rev:
    print("Palindrome number")
else:
    print("Not a palindrome number")
```

# Q3. Program to check whether a number is an Armstrong number or not

```python

num = int(input("Enter a number: "))
temp = num
total = 0
power = len(str(num))

while num > 0:
    digit = num % 10
    total = total + digit ** power
    num = num // 10

if temp == total:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

# Q4. Program to check whether a number is prime or not

```python

num = int(input("Enter a number: "))
count = 0

if num > 1:
    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```

# Q5. Program to check whether a number is a perfect number or not

```python

num = int(input("Enter a number: "))
total = 0

for i in range(1, num):
    if num % i == 0:
        total = total + i

if num > 0 and total == num:
    print("Perfect number")
else:
    print("Not a perfect number")
```

# Q6. Program to print prime numbers between 1 and 100

```python

for num in range(2, 101):
    count = 0

    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1

    if count == 2:
        print(num, end=" ")
```

```python
def details(*i,**j): #(*args,**kwargs)
    print("following are the guests for the meeting today:")
    for each in i:
        print(each)
    print("--------------\n\nDetails of your stay: \n--------------")
    for key , value in j.items():
        print(f"{key}:{value}")

Guests=["Amir","salman","shahrukh","kunal","pappu"]
Hotel_Details = {"city":"delhi", "location":"Meena bazaar", "hotel":"sunday", "room":80}

details(*Guests, **Hotel_Details)
```

```python
def details(x,*i,**j): #(parameter,*args,**kwargs)
    print("following are the guests for the meeting today:")
    for each in i:
        print(each)
    print("--------------\n\nDetails of your stay: \n--------------")
    for key , value in j.items():
        print(f"{key}:{value}")
    print(f"please note, following hotel contact no., in case:{x}")

Hotel_contact=9885356982
Guests=["Amir","salman","shahrukh","kunal","pappu"]
Hotel_Details = {"city":"delhi", "location":"Meena bazaar", "hotel":"sunday", "room":80}

details(Hotel_contact,*Guests, **Hotel_Details)
```

# Exception Handling

```python
x = 0
try:
    print(200/x)

except NameError:
    print("wrong keyword spelling, please check!")
    
except ZeroDivisionError:
    print("no. was divided by zero")
    
except Exception:
    print("Error Occured")

finally:
    print("program ends")
```
