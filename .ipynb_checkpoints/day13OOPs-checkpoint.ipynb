```python
class club_Mahindra():
    
    #constructor
    def __init__(self, cust_ID, Name, Membership, Subscription):
        self.cust_ID =cust_ID
        self.Name= Name
        self.Membership= Membership
        self.Subscription= Subscription
    #--------------------------

    def information(self):
        print(f"welcome {self.Name}, to club mahindra, your membership is (self.Membership)")

rekha = club_Mahindra(1001,"rekha patil", "gold", "yearly")
jaya = club_Mahindra(1001,"jaya joshi", "platinum", "monthly")
hema = club_Mahindra(1001,"hema kulkarni", "diamond", "half yearly")
```

```python
jaya.information()       #obj.method/func()
```
