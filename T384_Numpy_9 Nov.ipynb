```python
List1= [7,3,8,4,9,5,6,1,100,101,102,103]
len(List1)
# Convert it into an array of 4 rows and two columns
```

```python
arr1= np.array(List1)
print(arr1)
print("Dimension of arr1:",arr1.ndim)
print("--------------------------------")
arr2 = np.array(List1).reshape(4,3)
print(arr2)
print("Dimension of arr2:",arr2.ndim)

print("--------------------------------")
arr3 = np.array(List1).reshape(2,3,2)
print(arr3)
print("Dimension of arr3:",arr3.ndim)
```
