```python
class Jar:
    def __init__(self):
        self.candies = 100


class Withdraw:
    def execute(self, jar, amount):
        if amount <= jar.candies:
            jar.candies -= amount
            print("Withdraw:", amount)
        else:
            print("Not enough candies!")

        print("Remaining candies:", jar.candies)


class Deposit:
    def execute(self, jar, amount):
        jar.candies += amount
        print("Deposit:", amount)
        print("Total candies:", jar.candies)


jar = Jar()

user1 = Withdraw()
user2 = Deposit()
user3 = Withdraw()
user4 = Deposit()

amount = int(input("User 1: Enter candies to withdraw: "))
user1.execute(jar, amount)

amount = int(input("User 2: Enter candies to deposit: "))
user2.execute(jar, amount)

amount = int(input("User 3: Enter candies to withdraw: "))
user3.execute(jar, amount)

amount = int(input("User 4: Enter candies to deposit: "))
user4.execute(jar, amount)
        
```

# Encapsulation Abstraction

```python
## obj only access one class 
```

```python
## abstraction is focusing on the essential feature of an obj
```

```python
## abstraction define what an obj is.
## encapsulation define how an object protects and bundles its data/methods
```

```python
from abc import ABC , abstractmethod
# 1. define the abstract base class (the blueprint)
class Payment(ABC):
    
    @abstractmethod
    def process_payment(self,amount):
        print(amount)
#-------------------------

class CreditCard(Payment):
    def process_payment(self,amount):
        print(amount)

    def receipt(self,amount):
        print(amount)
```

```python
Ch_1 = CreditCard()
Ch_1.process_payment(10000) #parent obj accessing its own class method
```

```python
Child1= CreditCard()
Child1.process_payment(100) # child obj accessing the parent class method
```

```python
from abc import ABC , abstractmethod
# 1. define the abstract base class (the blueprint)
class Payment(ABC):
    
    @abstractmethod
    def process_payment(self,amount):
        """this method must be implementd by all subclass"""
        pass

    @abstractmethod
    def receipt(self,amount):
        """regular method can still exist in an abstract class"""
        print(f"receipt generated for ₹{amount}")
        
   
# 2. implement concerte subclass

class CreditCard(Payment):
    def process_payment(self,amount):
        print(f"processing credit card payment of ₹{amount} via scan & pay.")

    def receipt(self,amount):
        pass

class UPI(Payment):
    pass
    def process_payment(self,amount):
        print(f"processing UPI payment of ₹{amount} via scan & pay.")

    def receipt(self,amount):
        pass

cc = CreditCard()
cc.process_payment(5000)
cc.receipt(5000)

upi = UPI()
upi.process_payment(1200)
    
```
